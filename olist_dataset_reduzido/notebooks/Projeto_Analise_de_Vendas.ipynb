```python
# Trabalha com DataFrames (Tabelas) e Séries (Colunas)
import pandas as pd
```

```python
# Importação dos arquivos CSV
orders = pd.read_csv("https://raw.githubusercontent.com/marcosroquerosa/datasets/refs/heads/main/olist_dataset_reduzido/olist_orders_dataset.csv")
order_items = pd.read_csv("https://github.com/marcosroquerosa/datasets/raw/refs/heads/main/olist_dataset_reduzido/olist_order_items_dataset.csv")
customers = pd.read_csv("https://github.com/marcosroquerosa/datasets/raw/refs/heads/main/olist_dataset_reduzido/olist_customers_dataset.csv")
products = pd.read_csv("https://github.com/marcosroquerosa/datasets/raw/refs/heads/main/olist_dataset_reduzido/olist_products_dataset.csv")
sellers = pd.read_csv("https://github.com/marcosroquerosa/datasets/raw/refs/heads/main/olist_dataset_reduzido/olist_sellers_dataset.csv")
```

```python
# PROCV - ORDERS e da ORDER_ITEMS
# order_id, customer_id, data_pedido, status
dados_tratados = order_items.merge(orders[['order_id', 'order_purchase_timestamp',
                                           'order_status', 'customer_id']], on='order_id', how='left')
# merge é o PROCV
# how='left' mantem todas as linhas da tabela principal
```

```python
dados_tratados.dtypes
```

```python
dados_tratados.rename(columns={"order_purchase_timestamp":"data_pedido", "order_status":"status"}, inplace=True)
```

```python
dados_tratados.columns
```

```python
# Adicionar o estado do cliente - procv trouxe o estado
dados_tratados = dados_tratados.merge(
    customers[['customer_id', 'customer_state']], on='customer_id', how='left'
)
```

```python
dados_tratados.columns
```

```python
# Categoria do Produto - procv - categoria
# product_id	product_category_name
dados_tratados = dados_tratados.merge(
    products[['product_id','product_category_name']],
    on='product_id',
    how='left'
)
```

```python
dados_tratados.dtypes
```

```python
# Transformamos a data do pedido em formato datatiem
dados_tratados['data_pedido'] = pd.to_datetime(dados_tratados['data_pedido'])
dados_tratados.dtypes
```

```python
dados_tratados['mes'] = dados_tratados['data_pedido'].dt.month
dados_tratados['ano'] = dados_tratados['data_pedido'].dt.year
dados_tratados['mes_ano'] = dados_tratados['data_pedido'].dt.strftime('%m/%Y')
```

```python
dados_tratados.dtypes
```

```python
# Pedidos Válidos, atribui sim ou não.
# Inicialmente -Não para todos,
# posterior localização e atribuição de valor "Sim"
dados_tratados['pedido_valido'] = 'Não'
dados_tratados.loc[dados_tratados['status'] == 'delivered', 'pedido_valido'] = 'Sim'
```

```python
# Receita (preço + frete)
dados_tratados['receita'] = dados_tratados['price'] + dados_tratados['freight_value']
```

```python
dados_tratados.dtypes
```

```python
dados_tratados.drop(columns=['order_item_id', 'shipping_limit_date', ], inplace=True)
dados_tratados.columns
```

```python
dados_tratados.rename(columns={"product_category_name":"categoria", "customer_state":"estado_cliente"}, inplace=True)
```

```python
dados_tratados.columns
```

```python
# Reordenou as colunas do dataframe.

dados_tratados = dados_tratados[['order_id',
                                 'data_pedido',
                                 'status',
                                 'ano',
                                 'mes',
                                 'mes_ano',
                                 'pedido_valido',
                                 'customer_id',
                                 'estado_cliente',
                                 'product_id',
                                 'categoria',
                                 'seller_id',
                                 'price',
                                 'freight_value',
                                 'receita']]
```

```python
dados_tratados.columns
```

```python
# Output da tabela para um arquivo CSV
dados_tratados.to_csv('dados_tratados.csv', index=False)
```

```python
# Limpeza de Dados
# Remover dados nulos
print(dados_tratados.size)
dados_tratados.dropna()
print(dados_tratados.size)
```

```python
# Verificar registros duplicados
dados_tratados.duplicated().sum()
```

```python
# Remove itens duplicados
dados_tratados = dados_tratados.drop_duplicates()
```

```python
dados_tratados.size
```

```python
# Criar os KPIs
# Criar um novo dataframe somente com os pedidos válidos
validos = dados_tratados[dados_tratados['pedido_valido'] == 'Sim']
validos
```

```python
# Receita Total
receita_total = validos['receita'].sum()
print(f"Receita Total: R$ {receita_total:.2f}".replace(".",","))
```

```python
# Total de Pedidos
total_pedidos = validos['order_id'].nunique() # Contabiliza somente dados únicos, descarta repetidos)
print("Total de Pedidos: ", total_pedidos)
```

```python
# Ticket Médio
ticket_medio = receita_total / total_pedidos
print(f"Ticket Médio: R${ticket_medio:.2f}".replace(".",","))
```

```python
# Clientes Únicos
clientes_unicos = validos['customer_id'].nunique()
print("Clientes únicos: ", clientes_unicos)
```

```python
# Receita Média por cliente
receita_media_cliente = receita_total / clientes_unicos
print(f"Receita Média por cliente: R${receita_media_cliente:.2f}".replace(".",","))
```

```python
# Total de Itens Vendidos
total_itens = validos['product_id'].nunique()
print("Total de itens únicos vendidos: ", total_itens)
```

```python
# groupby - Agrupa receita por categoria, descendente, somente 5 primeiros
# KPI Receita por Categoria (Top 5)
top_categoria = (validos.groupby('categoria')['receita']
                 .sum() # Somar todos os itens de uma categoria
                 .sort_values(ascending=False)
                 .head(5)
                 )
top_categoria
```

```python
# Receita por Estado (Top 5)
top_estado = (validos.groupby('estado_cliente')['receita']
                 .sum() # Somar todos os itens de uma categoria
                 .sort_values(ascending=False)
                 .head(5)
                 )
top_estado
```

```python
# Receita por Vendedor (Top 5)
top_vendedor = (validos.groupby('seller_id')['receita']
                 .sum() # Somar todos os itens de uma categoria
                 .sort_values(ascending=False)
                 .head(5)
                 )
top_vendedor
```

```python
# Receita por Mês e Ano (Tendência)
receita_mes = (validos.groupby(['ano', 'mes'])['receita']
                 .sum() # Somar todos os itens de uma categoria
                 .reset_index()
                 .sort_values(['ano', 'mes'])
                 )
receita_mes
```

```python
# Receita por Mês e Ano (Tendência)
receita_mes_ano_desc = (validos.groupby(['ano', 'mes'])['receita']
                 .sum() # Somar todos os itens de uma categoria
                 .sort_values(ascending=False)
                 )
receita_mes_ano_desc
```
