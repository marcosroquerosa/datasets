```python
# Importando as bibliotecas
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
# Fazendo download de uma página
#Vamos acessar a página de notícias do IFPR.
url = "https://ifpr.edu.br/foz-do-iguacu/"
resposta = requests.get(url)
html = resposta.text
```

```python
#Criando o objeto BeautifulSoup
# Agora vamos transformar o HTML em um objeto analisável.
soup = BeautifulSoup(html, "html.parser")
```

```python
#Visualizando o HTML formatado
print(soup.prettify())
```

```python
# Encontrando elementos HTML
#O método find() retorna o primeiro elemento encontrado.
titulo = soup.find("h1")
print(titulo.text)
```

```python
# O método find_all() retorna uma lista de elementos.
titulos = soup.find_all("h2")

for t in titulos:
    print(t.text)
```

```python
#Buscando por classe CSS
#Muitos sites organizam notícias por classes CSS

noticias = soup.find_all("div", class_="row noticias-complementares")

for noticia in noticias:
    print(noticia.text)
```

```python
#Extraindo links das notícias
#Normalmente os títulos estão dentro de links.
links = soup.find_all("a")

for link in links:
    print(link.get("href"))
```

```python
conteudo = soup.find("main", id="conteudo")
if conteudo is None:
    print("Conteudo não encontrado")
else :
    noticias = conteudo.find_all("a")

    for noticia in noticias:
        titulo = noticia.text.strip()
        url = noticia.get("href")

        if titulo != None:
            print(titulo, url)
```

```python
dados = []

conteudo = soup.find("main", id="conteudo")
if conteudo is None:
    print("Conteudo não encontrado")
else :
    noticias = conteudo.find_all("a")

    for noticia in noticias:
        titulo_tag = noticia.find("h2", class_="complementares__titulo")
        titulo = titulo_tag.get_text(strip=True) if titulo_tag else ""

        url = noticia.get("href")

        descricao_p = noticia.find("p")
        descricao = descricao_p.get_text(strip=True) if descricao_p else ""

        if titulo != "":
            dados.append({
            "titulo": titulo,
            "descricao" : descricao,
            "link": url
        })
```

```python
df = pd.DataFrame(dados)

df
```

```python
# Salvando os dados
df.to_csv("noticias_ifpr.csv", index=False)
```

```python
# Extraindo as notícias
import time

def extrair_conteudo(url):
    try:
        time.sleep(1)

        response = requests.get(url)

        if response.status_code != 200:
            return ""

        soup = BeautifulSoup(response.text, "html.parser")

        # h1 (título da notícia)
        h1 = soup.find("h1")
        titulo = h1.get_text(strip=True) if h1 else ""

        # conteúdo (parágrafos)
        paragrafos = soup.find_all("p")

        texto = " ".join([p.get_text(strip=True) for p in paragrafos])

        return f"{titulo} {texto}"

    except Exception as e:
        print(f"Erro em {url}: {e}")
        return ""
```

```python
# Criando a coluna conteudo e extraindo dados das páginas.
df["conteudo"] = df["link"].apply(extrair_conteudo)

```

```python
# Outra maneira de fazer
# cria a coluna vazia antes (boa prática)
df["conteudo"] = ""
for index, row in df.iterrows():

    url = row["link"]
    # atualiza o dataframe
    df.at[index, "conteudo"] = extrair_conteudo(url) # Mais direto
    #df.loc[index, "conteudo"] = extrair_conteudo(url) # Mais pesado
df
```
