```python
# Tratar Datas
# (Provavelmente tem espações em branco e formato de data diferente, o mixed resolve mas pode converter errado)
# df['Release Date'] = pd.to_datetime(df['Release Date'], format="mixed")
# df['Created'] = pd.to_datetime(df['Created'], format="mixed")
# df['Modified'] = pd.to_datetime(df['Modified'], format="mixed")
```

```python
# 1. Limpeza básica Release Date
df["Release Date"] = df["Release Date"].astype(str).str.strip()

# 2. Conversão segura
df["Release Date"] = pd.to_datetime(df["Release Date"], errors="coerce")

# 3. Diagnóstico
erros = df[df["Release Date"].isnull()]

print("Linhas com erro:", len(erros))
erros.head()
```

```python
# 1. Limpeza básica Created
df["Created"] = df["Created"].astype(str).str.strip()

# 2. Conversão segura
df["Created"] = pd.to_datetime(df["Created"], errors="coerce")

# 3. Diagnóstico
erros = df[df["Created"].isnull()]

print("Linhas com erro:", len(erros))
erros.head()
```

```python
# 1. Limpeza básica Modified
df["Modified"] = df["Modified"].astype(str).str.strip()

# 2. Conversão segura
df["Modified"] = pd.to_datetime(df["Modified"], errors="coerce")

# 3. Diagnóstico
erros = df[df["Modified"].isnull()]

print("Linhas com erro:", len(erros))
erros.head()
```

```python
df.head()
```

```python
df["Bilheteria"] = (
    df["Description"]
    .str.extract(r"\$([\d,]+\.?\d*)")
    .replace(",", "", regex=True)
    .astype(float)
)
```

```python
df['Bilheteria']
```

```python
df.isnull().sum()
```

```python
df.dropna(subset=["IMDb Rating"])

#df["Bilheteria"].fillna(df["Bilheteria"].mean(), inplace=True)
#df["Bilheteria"] = df["Bilheteria"].fillna(df["Bilheteria"].mean())
```

```python
# Remover duplicatas
df.drop_duplicates(inplace=True)
```

```python
# Manipulação de Colunas
# Criar novas métricas
df["Bilheteria_por_minuto"] = df["Bilheteria"] / df["Runtime (mins)"]
```

```python
#Extrair ano da data
df["Ano_Lancamento"] = df["Release Date"].dt.year
```

```python
# Ordenação
df.sort_values("Bilheteria", ascending=False)
df.sort_values("IMDb Rating", ascending=False)
```

```python
# Agrupamento (GroupBy)
# Bilheteria média por ano
df.groupby("Year")["Bilheteria"].mean()
```

```python
# Número de filmes por diretor
df.groupby("Directors")["Title"].count()
```

```python
ranking = (
    df.groupby("Directors")
    .agg(
        total_filmes=("Title", "count"),
        filmes=("Title", lambda x: list(x))
    )
    .sort_values("total_filmes", ascending=False)
)

ranking.head()
```

```python
# Múltiplas métricas

df.groupby("Year").agg({
    "Bilheteria": ["sum", "mean"],
    "IMDb Rating": "mean"
})
```

```python
# Tabela Dinâmica
# Bilheteria por ano

pd.pivot_table(
    df,
    values="Bilheteria",
    index="Year",
    aggfunc="sum"
)
```

```python
# Nota média por gênero (cuidado: múltiplos gêneros)
# Explodir gêneros:
df_exploded = df.assign(Genre=df["Genres"].str.split(", ")).explode("Genre")

pd.pivot_table(
    df_exploded,
    values="IMDb Rating",
    index="Genre",
    aggfunc="mean"
).sort_values(by="IMDb Rating", ascending=False)
```
